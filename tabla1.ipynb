```python
import re
patron = r'\n(\w+\s*\w+\s*\w*)\s+(\d+)\s+(\d+.\d+)\s+(\d+)\s+(\d+.\d+)' 
datos = re.findall(patron, texto)


etas = pd.DataFrame(datos, columns=['Area de Salud', 'Casos 2017', 'Tasas 2017', 'Casos 2018', 'Tasas 2018'])
etas
```
