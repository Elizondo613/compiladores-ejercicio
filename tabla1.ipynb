```python
import pandas as pd
from PyPDF2 import PdfReader
rutaArchivo = 'C:/Users/Elizondo/Desktop/Universidad/Compi/EJ-analizador/Analisis Anual 2018 ETAS.pdf'
documento = PdfReader(rutaArchivo)
dir(documento)
```

```python
len(documento.pages)
pagina = documento.pages[2]
texto = pagina.extract_text()
texto.split('\n')
```

```python
import re
patron = r'\n(\w+\s*\w+\s*\w*)\s+(\d+)\s+(\d+.\d+)\s+(\d+)\s+(\d+.\d+)' 
datos = re.findall(patron, texto)


etas = pd.DataFrame(datos, columns=['Area de Salud', 'Casos 2017', 'Tasas 2017', 'Casos 2018', 'Tasas 2018'])
etas.info()
```

```python
etas['Casos 2017'] = etas['Casos 2017'].astype('int64')
etas['Tasas 2017'] = etas['Tasas 2017'].astype('float64')
etas['Casos 2018'] = etas['Casos 2018'].astype('int64')
etas['Tasas 2018'] = etas['Tasas 2018'].astype('float64')
etas.info()
```

```python
#Cuando se afecta al conjunto de columnas
etas = etas.astype({'Casos 2017': 'int64', 'Tasas 2017': 'float64', 'Casos 2018': 'int64', 'Tasas 2018': 'float64'})
```
