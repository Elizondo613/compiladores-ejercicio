```python
import pandas as pd
from PyPDF2 import PdfReader

rutaArchivo = 'C:/Users/Elizondo/Desktop/Universidad/Compi/EJ-analizador/Analisis Anual 2018 ETAS.pdf'
documento = PdfReader(rutaArchivo)

len(documento.pages)
pagina = documento.pages[4]
texto = pagina.extract_text()
texto
```

```python
import re
patron = r'\n\s*(\d+)\s*(\w+\s*\w+\s*\w+)\s*(\w+\s*\w+\s*)\s*(\w+\s*\w+\s*\w+\s*\w+\s*\w+\s*\w+)\s*(\d+)' 
datos = re.findall(patron, texto)


etas = pd.DataFrame(datos, columns=['SE', 'Area de salud', 'ETA Reportada', 'Fuente de contagio', 'Numero de casos'])
etas
```
